## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Fetch the raw CSV directly from your GitHub repo
url = "https://raw.githubusercontent.com/sheby-me/FLYRANK-Onboarding/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Display total rows and columns to confirm it worked
print(df.shape)
print(df.head())
print(df.columns) # Added to show all column names
missing_position = len(df[df['avg_position'] == 0])
total_rows = len(df)

# Calculate the base rate by checking how many rows have a 'down' trend
decline_rate = (df['trend_direction'] == 'down').mean() * 100

print(f"Total content items: {total_rows:,}")
print(f"Items with missing position data (avg_position=0): {missing_position:,}")
print(f"Base rate of declining content: {decline_rate:.1f}%")
```
